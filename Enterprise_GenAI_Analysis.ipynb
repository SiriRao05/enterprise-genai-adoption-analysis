```python
from scipy.stats import chi2_contingency
table = pd.crosstab(
    df["Industry"],
    df["GenAI Tool"]
)
table
chi2, p_value, degrees_of_freedom, expected = chi2_contingency(table)
print("Chi-square statistic:", chi2)
print("Degrees of freedom:", degrees_of_freedom)
print("p-value:", p_value)

if p_value < 0.05:
    print("Result: There is a statistically significant association between industry and GenAI tool.")
else:
    print("Result: There is not enough evidence of an association between industry and GenAI tool.")
```
